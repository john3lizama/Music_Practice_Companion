```python
song_y, song_sr = librosa.load("./temp_folder/vocadito_2.wav", sr=44100)
scale_y2, scale_sr2 = librosa.load("./temp_folder/f3_scales_straight_a.wav", sr=44100)
librosa.load()
```

```python
print("Sample rate:", song_sr)
print("Duration (seconds):", len(song_y)/song_sr)

print("Sample rate:", scale_sr2)
print("Duration (seconds):", len(scale_y2)/scale_sr2)
```

```python
plt.figure()
librosa.display.waveshow(song_y, sr=song_sr)
plt.title("Waveform of vocadito_2.wav")
plt.show()

plt.figure()
librosa.display.waveshow(scale_y2, sr=scale_sr2)
plt.title("Waveform of f3_scales_straight_a.wav")
plt.show()

plt.figure()
librosa.display.specshow(librosa.amplitude_to_db(librosa.stft(song_y), ref=np.max), sr=song_sr, x_axis='time', y_axis='log')
plt.colorbar(format='%+2.0f dB')
plt.title("Spectrogram of vocadito_2.wav")
plt.show()
```

```python
f0, voiced_flag, voiced_probs = librosa.pyin(
    song_y,
    fmin=librosa.note_to_hz("C2"),
    fmax=librosa.note_to_hz("C7")
)

print("Average pitch (Hz):", np.nanmean(f0))

tempo, _ = librosa.beat.beat_track(y=song_y, sr=song_sr)
print("Estimated tempo:", tempo)
```
